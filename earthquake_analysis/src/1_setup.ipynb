```python
%sql
create connection if not exists earthquakeConn
type http
options (
    host = "https://earthquake.usgs.gov",
    port = 443,
    base_path = "/earthquakes/feed/v1.0/",
    bearer_token = "na"
)
```

```python
%sql
use catalog earthquake;
use schema bronze;
create volume if not exists earthquake_data;
```

```python
from databricks.sdk import WorkspaceClient
w=WorkspaceClient()
conn=w.connections.get("earthquakeConn")
```

```python
url=f"{conn.options["host"]}{conn.options["base_path"]}summary/all_day.geojson"
import requests
import json
response = requests.get(url)
data = response.json()
dbutils.fs.put("/Volumes/earthquake/bronze/earthquake_data/earthquake_data.json",json.dumps(data),overwrite=True)
```
